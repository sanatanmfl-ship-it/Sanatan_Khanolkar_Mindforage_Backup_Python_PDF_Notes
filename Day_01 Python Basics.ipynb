## 1. What is Python?

Python is a high-level, general-purpose programming language with readable syntax.

- **High-level:** code uses concepts such as lists and dictionaries rather than requiring manual memory management.
- **Interpreted workflow:** the Python runtime executes code and reports errors.
- **Dynamically typed:** names can refer to objects of different types at different times.
- **Multi-paradigm:** supports procedural, object-oriented, and functional programming.

### Why Python?
- Readable syntax helps beginners focus on problem-solving.
- Large standard library and third-party ecosystem.
- Useful for automation, web backends, data analysis, AI/ML, testing, and scripting.
- Trade-off: pure Python may be slower than compiled languages for CPU-intensive tasks.

### Where is Python used?
| Area | Example |
|---|---|
| Automation | Rename files or generate reports |
| Web development | Build APIs and server-side applications |
| Data analysis | Clean and summarize datasets |
| AI / machine learning | Experiment with models |
| Testing | Automate checks for software |
| Education | Teach programming concepts |

## 3. Comments, statements, and indentation

- A comment begins with `#`; it explains code and is not executed.
- Python uses indentation to group statements into blocks.
- Use four spaces per indentation level.
- A colon (`:`) introduces a block after statements such as `if`, `for`, `def`, and `class`.

```python
# This is a comment
student_name = "Asha"

if student_name:
    print(f"Hello, {student_name}!")
```

## 4. Variables and assignment

A variable is a **name that refers to an object**. The `=` operator assigns a value to a name; it does not test equality.

Good names describe the purpose of the value. Python names are case-sensitive: `score` and `Score` are different.

```python
student_name = "Asha"
age = 19
temperature = 28.5
is_enrolled = True

print(student_name)
print(age)
print(temperature)
print(is_enrolled)
```

## 5. Built-in data types

| Type | Example | Typical use |
|---|---|---|
| `int` | `42` | Whole numbers |
| `float` | `3.14` | Decimal numbers |
| `str` | `"Python"` | Text |
| `bool` | `True` | Logical conditions |

```python
price = 49.50
language = "Python"
is_ready = False
not_set = None

print(type(price))
print(type(language))
print(type(is_ready))
print(type(not_set))
```

## 6. Keywords and identifiers

**Identifier:** a name chosen for a variable, function, class, or other object.

Rules:
- May contain letters, digits, and underscores.
- Cannot begin with a digit.
- Cannot contain spaces or hyphens.
- Is case-sensitive.
- Cannot be a reserved keyword.

**Keyword:** a reserved word with a special meaning, such as `if`, `for`, `class`, `return`, `True`, and `None`.

Examples:
- Valid: `total_marks`, `_count`, `student2`
- Invalid: `2score`, `student-name`, `class`

Some words such as `match` and `case` are soft keywords used in specific syntax contexts.

```python
# Valid identifiers
total_marks = 95
student2 = "Ravi"
_count = 3

print(total_marks, student2, _count)

# Uncomment the next line to see a SyntaxError:
# class = "Python" 
```

## 7. Operators

Arithmetic operators: `+`, `-`, `*`, `/`, `//`, `%`, `**`  
Comparison operators: `==`, `!=`, `<`, `<=`, `>`, `>=`  
Logical operators: `and`, `or`, `not`

Remember: `=` assigns a value, while `==` compares values.

```python
a, b = 7, 2

print("Addition:", a + b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** b)
print("Is a greater than b?", a > b)
print("Are they equal?", a == b)
```

## 8. Type conversion

Convert values when the operation requires a different type. Some conversions can fail, so validate user input in real programs.

```python
age_text = "19"
age = int(age_text)
height = float("1.72")
message = "Age: " + str(age)

print(age + 1)
print(height)
print(message)

# Try int("hello") to observe a ValueError (run separately).
```

## 9. Input and formatted output

`input()` always returns a string. Convert it before doing numeric calculations. An f-string inserts expressions into text.

```python
name = input("Your name: ").strip()
age_text = input("Your age: ").strip()

try:
    age = int(age_text)
except ValueError:
    print("Please enter your age as a whole number.")
else:
    print(f"Hello, {name or 'student'}!")
    print(f"Next year you will be {age + 1}.")
```
