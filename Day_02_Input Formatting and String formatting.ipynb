## 3. Convert input: `int()` and `float()`

Because `input()` returns text, convert it before numeric calculations.

- `int("25")` produces integer `25`.
- `float("25.5")` produces decimal value `25.5`.
- `int("25.5")` raises `ValueError`; the text is not an integer literal.


````python
age = int(input("Enter your age: "))
print("Next year you will be", age + 15)
````

````python
price = float(input("Enter item price: "))
quantity = int(input("Enter quantity: "))
total = price * quantity
print("Total:", total)
````

### Handle invalid numeric input

If a user types letters where a number is expected, `int()` or `float()` can raise `ValueError`. Use `try` / `except` to handle it gracefully.

## 5. f-strings: readable formatted output

An f-string starts with `f` before the quote. Put a variable or expression inside curly braces `{}`.

```python

age = 19
print(f"{name} is {age} years old.")
```

f-strings are usually the clearest option for inserting values into messages.


````python
name = "Ram"
age = 35
print(f"Hi , my name is {name} and tomorrow i will be {age} years old")
````

## 6. Quick Reference: Number Formatting in f-Strings

Use a **format specifier** inside an f-string to control how a number appears on the screen. The format does not change the original value stored in the variable.

### Format specifiers

| Format specifier | Purpose | Example output |
|---|---|---:|
| `:.1f` | Display 1 digit after the decimal point | `85.7` |
| `:.2f` | Display 2 digits after the decimal point | `85.68` |
| `:.0f` | Display no digits after the decimal point | `86` |
| `:,` | Add thousands separators | `1,250,000` |
| `:,.2f` | Add thousands separators and display 2 decimal places | `1,234,567.89` |
| `:.2%` | Display a proportion as a percentage with 2 decimal places | `85.67%` |

### Example

```python
marks = 85.6789
amount = 1234567.891
score = 0.8567

print(f"One decimal place: {marks:.1f}")
print(f"Two decimal places: {marks:.2f}")
print(f"No decimal places: {marks:.0f}")
print(f"Amount with commas: {amount:,}")
print(f"Amount with commas and 2 decimals: {amount:,.2f}")
print(f"Percentage: {score:.2%}")
```

Expected output:

```text
One decimal place: 85.7
Two decimal places: 85.68
No decimal places: 86
Amount with commas: 1,234,567.891
Amount with commas and 2 decimals: 1,234,567.89
Percentage: 85.67%
```

**Remember:** `.1f` and `.2f` control the number of decimal places shown. Formatting rounds the displayed result; it does not permanently modify the variable.

````python
student_name = "Asha"
age = 19
marks = 87.45667889292033839923

print(f"Student: {student_name}")
print(f"Age: {age}")
print(f"Marks: {marks:.3f}")
print(f"{student_name} is {age} years old and scored {marks:.0f} marks.")
````

## 7. Basic string operations

A string (`str`) is text enclosed in quotes. Strings are immutable: string methods return a new string rather than changing the original string in place.

Common operations:
- `+` joins strings.
- `*` repeats a string.
- `len()` counts characters.
- Indexing accesses a character; indexing starts at `0`.


````python
first_name = "Asha"
last_name = "Patil"
full_name = first_name +" "+ last_name

print(full_name)
print("Hi! " * 3)
print("Length:", len(full_name))
print("Second character of first name:", first_name[3])
print("Third character of last name:", last_name[-3])
print("Third character of last name:", last_name[2])

````

## 8. Useful string methods

| Method | Purpose |
|---|---|
| `.lower()` | Convert to lowercase |
| `.upper()` | Convert to uppercase |
| `.title()` | Capitalize the start of words |
| `.strip()` | Remove surrounding whitespace |
| `.replace(old, new)` | Replace matching text |
| `.split()` | Split text into a list |


````python
message = "            welcome to python           "

print(message.strip())
print(message.lstrip().lower())
print(message.rstrip().lower())
print(message.strip().upper())
print("hello student".title())
print("I like Java".replace("Java", "Python"))
print("red,green,blue".split(","))
````

## 9. Mini project: Student information and marks receipt

This mini project combines input, type conversion, string cleanup, and formatted output.


````python
name = input("Student name: ").strip()
subject = input("Subject: ").strip()
marks = float(input("Marks obtained: "))
total_marks = float(input("Total marks: "))


percentage = marks / total_marks
print("\n--- Student Report ---")
print(f"Name       : {name}")
print(f"Subject    : {subject}")
print(f"Marks      : {marks:.1f} / {total_marks:.1f}")
print(f"Percentage : {percentage:.1%}")
````

## 10. Common errors and good practices

- **`NameError`**: using a variable name that has not been defined.
- **`ValueError`**: trying `int("hello")` or `float("abc")`.
- **`TypeError`**: combining incompatible types, such as `"Age: " + 19` without converting `19` to text.
- **Case sensitivity**: `input()` is correct; `Input()` is a different name.
- Use `.strip()` for text fields where leading/trailing spaces are not wanted.
- Use f-strings for readable output.
- Use `float()` for decimal input and `int()` for whole-number input.
